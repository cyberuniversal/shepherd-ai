Roadmap role: Week 4. Literature-driven rule: consume validated intents and grounded targets; produce inspectable task sequences or graphs. This notebook runs the deterministic Week 4 planning slice. It does not schedule drones, optimize routes, execute simulation, or validate safety.

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / 'src').exists() and (ROOT / 'shepherd-ai').exists():
    ROOT = ROOT / 'shepherd-ai'
sys.path.insert(0, str(ROOT / 'src'))

MAP_PATH = ROOT / 'datasets' / 'maps' / 'shepherd_test_map_v1.csv'
HUMAN_GROUNDING_BENCHMARK = ROOT / 'datasets' / 'maps' / 'human_grounding_benchmark_v1.jsonl'
WEEK4_PLANNING_CASES = ROOT / 'datasets' / 'maps' / 'week4_planning_cases_v1.jsonl'

print('root', ROOT)
print('map exists', MAP_PATH.exists())
print('human grounding benchmark exists', HUMAN_GROUNDING_BENCHMARK.exists())
print('week4 planning cases exist', WEEK4_PLANNING_CASES.exists())
```

```python
from shepherd_ai.grounding import ground_intent, load_map_locations
from shepherd_ai.intent import parse_intent
from shepherd_ai.mission_planning import plan_grounded_mission, validate_mission_plan

locations = load_map_locations(MAP_PATH)
intent = parse_intent('Scan the crops in the north field.')
grounded = ground_intent(intent, locations)
plan = plan_grounded_mission(grounded)
{'plan': plan.to_dict(), 'validation': validate_mission_plan(plan).to_dict()}
```
